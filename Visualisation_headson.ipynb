```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'C:\Users\sujal\OneDrive\Documents\Intellipet\Visualisation\Student_Performance_Dataset.csv')
```

```python
df
```

```python
# 1. distribution of maths scores
         
               # using matplotib
plt.hist(df['Math_Score'], bins = 10, edgecolor = 'black')
plt.xlabel('Maths_score')
plt.ylabel('count')
plt.title('distribution of maths score')
plt.show()
```

```python
# using seaborn

sns.histplot(df['Math_Score'])
plt.title("Distribution og maths score")
plt.show()
```

```python
# 2. Gender wise average final percentage

avg_score = df.groupby('Gender')['Final_Percentage'].mean()
avg_score
```

```python
plt.bar(avg_score.index, avg_score.values)
plt.title("Average final persentage by gender")
plt.show()
```

```python
# using seaborn

sns.barplot(x = 'Gender', y = 'Final_Percentage', data = df)
plt.title('Average final percentage by gender')
plt.show()
```

```python
# 3. relationship between study hours and final percentage

plt.scatter(df['Study_Hours_Per_Day'],df['Final_Percentage'])
plt.show
```

```python
# using seaborn

sns.scatterplot(x = 'Study_Hours_Per_Day',y = 'Final_Percentage', data = df)
plt.show()
```

```python
# 4. count of student by performance level
count = df['Performance_Level'].value_counts()
```

```python
count
```

```python
plt.bar(count.index, count.values)
plt.title('Performance level distrubution')
plt.show()
```

```python
sns.countplot(x ='Performance_Level', data = df)
plt.title('Performance level distrubution')
plt.show()
```

```python
# 5. what is distribution of pass or fail students

pass_fail_student = df['Pass_Fail'].value_counts()
pass_fail_student
```

```python
plt.pie(pass_fail_student.values, labels = pass_fail_student.index,
         autopct = '%1.1f%%', shadow = True)
plt.title('Pass vs Fail destribution')
plt.show()
```

```python
# 6. create a 2*2 subplot dashbord showing distribution of math score, pass vs fail
# count, study hours vs final percentage, pass vs fail distribution

plt.subplot(2,2,1)
plt.hist(df['Math_Score'], bins = 10, color = 'pink', edgecolor = 'black')
plt.title('math score distribution')
plt.xlabel('math score')
plt.ylabel('count')

plt.subplot(2,2,2)
pass_count = df['Pass_Fail'].value_counts()
plt.bar(pass_count.index, pass_count.values, color = ['green', 'red'])
plt.title('pass vs fail')
plt.xlabel('pass / fail')
plt.ylabel('count')

plt.subplot(2,2,3)
plt.scatter(df['Study_Hours_Per_Day'], df['Final_Percentage'], color = 'purple')
plt.title('study hours vs persentage')
plt.xlabel('study hours vs perrcentage')
plt.ylabel('Final %')

plt.subplot(2,2,4)
plt.pie(pass_fail_student.values, labels = pass_fail_student.index,
         autopct = '%1.1f%%', shadow = True)
plt.title('Pass vs Fail destribution')
plt.tight_layout()
plt.show()
```
