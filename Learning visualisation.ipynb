```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

a = np.array([0,6])
b = np.array([0,150])
```

```python
plt.plot(a,b)
plt.show()

# plot is a function that is used to creat line plt from one point to another point
```

```python
plt.plot(a,b,'o')
plt.show()
```

```python
x= [1,2,3,5,6,7]
y= [2,5,6,7,8,9]

plt.plot(x,y,marker='o',ms =10,mec = 'black',mfc = 'y',linestyle = 'dashed',color ='r',
         linewidth = '2')
plt.show()

# marker size => ms
# marker edge color => mec
# marker face color => mfc
# linestyle => linestyle (dashed, dotted, dash -dot, solid etc)
# line color => color
# linewidth => linewidth
```

```python
a1 = np.array([3,8,1,10,14])
a2 = np.array([6,2,7,11,12])

plt.plot(a1)
plt.plot(a2)

plt.xlabel('Study hrs')
plt.ylabel('Marks')
plt.title('Academics', loc = 'left')
plt.grid(axis = 'y')
plt.show()
```
